# Building prompts dynamically

```python
import os
from dotenv import load_dotenv

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate


# --------------------------------------------------
# 1. Load environment variables
# --------------------------------------------------

load_dotenv()


# --------------------------------------------------
# 2. Create the LLM
# --------------------------------------------------

llm = ChatOpenAI(
    model="gpt-5-mini",
    api_key=os.getenv("OPENAI_API_KEY")
)


# --------------------------------------------------
# 3. Create a PromptTemplate
# --------------------------------------------------

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple English with an example."
)


# --------------------------------------------------
# 4. Provide the value for the variable
# --------------------------------------------------

ready_prompt = prompt.invoke({
    "topic": "RAG"
})


# --------------------------------------------------
# 5. Send the prompt to the LLM
# --------------------------------------------------

response = llm.invoke(ready_prompt)


# --------------------------------------------------
# 6. Display the response
# --------------------------------------------------

print(response.content)
```
