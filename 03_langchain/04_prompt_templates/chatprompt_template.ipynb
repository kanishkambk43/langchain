## if we want to involve system message or any message classes into the prompt that we are building then we use chatprompttemplate

```python
import os
from dotenv import load_dotenv

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate


# --------------------------------------------------
# 1. Load environment variables
# --------------------------------------------------

load_dotenv()


# --------------------------------------------------
# 2. Create the LLM
# --------------------------------------------------

llm = ChatOpenAI(
    model="gpt-5-mini",
    api_key=os.getenv("OPENAI_API_KEY")
)


# --------------------------------------------------
# 3. Create a ChatPromptTemplate
# --------------------------------------------------

chat_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful teacher. Explain concepts in simple English."
    ),
    (
        "human",
        "Explain {topic} with a simple example."
    )
])


# --------------------------------------------------
# 4. Provide the value for the variable
# --------------------------------------------------

ready_messages = chat_prompt.invoke({
    "topic": "RAG"
})


# --------------------------------------------------
# 5. Send the messages to the LLM
# --------------------------------------------------

response = llm.invoke(ready_messages)


# --------------------------------------------------
# 6. Display the response
# --------------------------------------------------

print(response.content)
```
